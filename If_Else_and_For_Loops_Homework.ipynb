## Optional challenge: shopping receipt
Use a loop to total `prices`. If the subtotal is at least 50, apply a 10% discount; otherwise, apply no discount. Print the subtotal, discount amount, and final total with two decimal places. You may use an f-string such as `f"{value:.2f}"`.

**Test cases:** supplied list → subtotal `57.50`, discount `5.75`, final total `51.75`; `[10.00, 15.00]` → final total `25.00`; `[]` → final total `0.00`.


```python
prices = [12.50, 20.00, 25.00]
subtotal = 0
for price in prices:
    subtotal = subtotal + price
if subtotal >= 50:
    discount = subtotal * 0.10
else:
      discount = 0

final_total = subtotal - discount
print(f"Subtotal: {subtotal:.2f}")
print(f"Discount: {discount:.2f}")
print(f"Final total: {final_total:.2f}")


# Write your code here.
```
